```python
print("junction_control = -1, what junction_detail do they have?")
mask = df["junction_control"] == -1
print(df[mask]["junction_detail"].value_counts().head(5))
```

44,397 out of 44,977 rows with junction_control = -1 have junction_detail = 0 meaning no junction at all. This is not a missing data. We have similar more features with -1 values, will replace unknown with sensible values.

- carriageway_hazards - replace with 0
- any_skidding - replace with 0
- avg_vehicle_age - replace with median
- road_surface_conditions - replace with mode
- avg_casualty_age - replace with median

```python
road_surface_mode = df[df["road_surface_conditions"] != -1]["road_surface_conditions"].mode()[0]
df["road_surface_conditions"] = df["road_surface_conditions"].replace(-1, road_surface_mode)

df["carriageway_hazards"] = df["carriageway_hazards"].replace(-1, 0)
df["carriageway_hazards"] = df["carriageway_hazards"].replace(99, 0)

vehicle_type_mode = df[df["vehicle_type_mode"] != -1]["vehicle_type_mode"].mode()[0]
df["vehicle_type_mode"] = df["vehicle_type_mode"].replace(-1, vehicle_type_mode)

df["any_skidding"] = df["any_skidding"].replace(-1, 0)

vehicle_age_median = df[df["avg_vehicle_age"] != -1]["avg_vehicle_age"].median()
df["avg_vehicle_age"] = df["avg_vehicle_age"].replace(-1, vehicle_age_median)

casualty_age_median = df[df["avg_casualty_age"] != -1]["avg_casualty_age"].median()
df["avg_casualty_age"] = df["avg_casualty_age"].replace(-1, casualty_age_median)

print("Sentinel values cleaned.")

print("Remaining -1 values:")
for col in df.columns:
    count = (df[col] == -1).sum()
    if count > 0:
        print(f"{col}:{count}")

print("carriageway_hazards max:", df["carriageway_hazards"].max())
```

### Fatal Accident Rate by Speed Limit

```python
# what percentage of accidents at each speed limit are fatal?
speed_limits = [20, 30, 40, 50, 60, 70]
fatal_rates = []

for speed in speed_limits:
    subset = df[df["speed_limit"] == speed]
    fatal_count = (subset["collision_severity"] == 1).sum()
    total_count = len(subset)
    fatal_pct = (fatal_count / total_count) * 100
    fatal_rates.append(round(fatal_pct, 2))
    print(f"{speed}mph - {fatal_count} fatal out of {total_count} = {fatal_pct:.2f}%")

plt.figure(figsize=(6, 4))
plt.bar([str(s) for s in speed_limits], fatal_rates, color="#1f77b4")
plt.title("Fatal Accident Rate by Speed Limit")
plt.xlabel("Speed Limit (mph)")
plt.ylabel("Fatal Accidents (%)")

for i, val in enumerate(speed_fatal.values):
    plt.text(i, val + 0.05, f"{val:.1f}%", ha="center", fontsize=9)

plt.tight_layout()
plt.savefig("../reports/figures/speed_vs_fatal.png", dpi=100, bbox_inches="tight")
plt.show()
```

### Light conditions vs fatal rate

```python
light_labels = {1: "Daylight", 4: "Dark-lit", 5: "Dark-unlit", 6: "Dark-no light", 7: "Dark-unknown"}
light_values = sorted(light_labels.keys())

fatal_rates_light = []

for val in light_values:
    subset = df[df["light_conditions"] == val]
    fatal_count = (subset["collision_severity"] == 1).sum()
    total_count = len(subset)
    fatal_pct = (fatal_count / total_count) * 100
    fatal_rates_light.append(round(fatal_pct, 2))
    print(f"{light_labels[val]} - {fatal_pct:.2f}%")

plt.figure(figsize=(7, 4))
plt.bar([light_labels[v] for v in light_values], fatal_rates_light, color="#ff7f0e")
plt.title("Fatal Accident Rate by Light Conditions")
plt.xlabel("Light Condition")
plt.ylabel("Fatal Accidents (%)")
plt.xticks(rotation=15)

for i, val in enumerate(fatal_rates_light):
    plt.text(i, val + 0.05, f"{val:.1f}%", ha="center", fontsize=9)
    
plt.tight_layout()
plt.savefig("../reports/figures/light_vs_fatal.png", dpi=100, bbox_inches="tight")
plt.show()
```

### Urban vs Rural

```python
area_values = sorted(df["urban_or_rural_area"].unique())
print(area_values)
```

```python
area_labels = {1: "Urban", 2: "Rural", 3: "Unallocated"}

fatal_rates_area = []

for val in area_values:
    subset = df[df["urban_or_rural_area"] == val]
    fatal_count = (subset["collision_severity"] == 1).sum()
    total_count = len(subset)
    fatal_pct = (fatal_count / total_count) * 100
    fatal_rates_area.append(round(fatal_pct, 2))
    print(f"{area_labels[val]} - {fatal_count} fatal out of {total_count} = {fatal_pct:.2f}%")

plt.figure(figsize=(5, 4))
plt.bar([area_labels[v] for v in area_values], fatal_rates_area, color="#2ca02c")
plt.title("Fatal Accident Rate by Area Type")
plt.xlabel("Area Type")
plt.ylabel("Fatal Accidents (%)")

for i, val in enumerate(fatal_rates_area):
    plt.text(i, val + 0.05, f"{val:.1f}%", ha="center", fontsize=9)

plt.tight_layout()
plt.savefig("../reports/figures/urban_rural_vs_fatal.png", dpi=100, bbox_inches="tight")
plt.show()
```

### Number of casualties vs fatal rate

```python
casualty_counts = sorted(df["number_of_casualties"].unique())[:8]
fatal_rates = []

for n in casualty_counts:
    subset = df[df["number_of_casualties"] == n]
    fatal_rates.append((subset["collision_severity"] == 1).sum() / len(subset) * 100)

plt.figure(figsize=(6, 4))
plt.bar([str(n) for n in casualty_counts], fatal_rates, color="#9467bd")
plt.title("Fatal Rate by Number of Casualties")
plt.xlabel("Number of Casualties")
plt.ylabel("Fatal Accidents (%)")

for i, val in enumerate(fatal_rates):
    plt.text(i, val + 0.05, f"{val:.1f}%", ha="center", fontsize=9)
    
plt.tight_layout()
plt.savefig("../reports/figures/casualties_vs_fatal.png", dpi=100, bbox_inches="tight")
plt.show()
```

```python
corr_cols = ["collision_severity", "speed_limit", "number_of_vehicles",
             "number_of_casualties", "avg_vehicle_age", "light_conditions",
             "urban_or_rural_area"]

corr = df[corr_cols].corr()

plt.figure(figsize=(7, 5))
plt.imshow(corr, cmap="coolwarm", aspect="auto")
plt.colorbar()

for i in range(len(corr_cols)):
    for j in range(len(corr_cols)):
        plt.text(j, i, f"{corr.iloc[i,j]:.2f}", ha="center", va="center", fontsize=9, color="black")
                 
plt.xticks(range(len(corr_cols)), corr_cols, rotation=45, ha="right", fontsize=8)
plt.yticks(range(len(corr_cols)), corr_cols, fontsize=8)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.savefig("../reports/figures/correlation_heatmap.png", dpi=100, bbox_inches="tight")
plt.show()
```
