```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/processed/accidents_clean.csv")
print(df.shape)
print(df.head())
```

```python
# convert time string to minutes since midnight
df["time_minutes"] = pd.to_datetime(df["time"], format="%H:%M").dt.hour * 60 + \
                     pd.to_datetime(df["time"], format="%H:%M").dt.minute

# convert to cyclical features
df["time_sin"] = np.sin(2 * np.pi * df["time_minutes"] / 1440)
df["time_cos"] = np.cos(2 * np.pi * df["time_minutes"] / 1440)

# drop original time columns 
df = df.drop(columns=["time", "time_minutes"])

print("New columns added: time_sin, time_cos")
print(df[["time_sin", "time_cos"]].head())
```

```python
df["day_sin"] = np.sin(2 * np.pi * df["day_of_week"] / 7)
df["day_cos"] = np.cos(2 * np.pi * df["day_of_week"] / 7)

df = df.drop(columns=["day_of_week"])

print(df[["day_sin", "day_cos"]].head())
```

```python
print(df.shape)
print(df.columns.tolist())
```

```python
df["speed_x_area"] = df["speed_limit"] * df["urban_or_rural_area"]

print("Interaction feature added: speed_x_area")
print(df["speed_x_area"].describe().round(2))
```

```python
leaking_cols = ["worst_casualty_severity", "any_pedestrian", "avg_casualty_age"]
df = df.drop(columns=leaking_cols)

print("Leaking columns removed.")
print("Final shape:", df.shape)
print("Final columns:", df.columns.tolist())
```

```python
df.to_csv("../data/processed/accidents_features.csv", index=False)
print("Saved:", df.shape)
```
